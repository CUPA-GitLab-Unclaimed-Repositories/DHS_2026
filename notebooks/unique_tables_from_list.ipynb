```python
def identify_unique_table_names(table_list):
    """
    Returns a list of unique table names, preserving the original order.
    """
    seen=set()
    unique_table_name = []
    for table in table_list:
        table_lower = table.lower()
        if table_lower not in seen:
            seen.add(table_lower)
            unique_table_name.append(table)
    for table_name in unique_table_name:
        print(table_name)
    return unique_table_name
```
