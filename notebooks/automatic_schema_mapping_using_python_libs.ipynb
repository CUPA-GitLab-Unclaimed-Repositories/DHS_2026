```python
import pandas as pd
from fuzzywuzzy import process
```

```python
sas_db = pd.read_csv(f'../data/SessionItemMarks_schema.csv')  
digital_db = pd.read_csv(f'../data/digital_schema.csv')  

sas_columns = sas_db['column_name'].tolist()
digital_columns = digital_db['column_name'].tolist()

# Map columns using fuzzy matching
mapping = []
for col in sas_columns:
    match, score = process.extractOne(col, digital_columns)
    mapping.append({'sas_db_column': col, 'digital_db_column': match, 'match_score': score})

# Convert mapping to DataFrame and save
mapping_df = pd.DataFrame(mapping)
# mapping_df.to_csv('column_mapping.csv', index=False)

# print("Column mapping saved to column_mapping.csv")
```
